```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from astropy.table import Table
from scipy.optimize import curve_fit
import random

# -- 3: Review --
# -- 3.1 Plotting an Astronomical Image --
# 1)
data = np.loadtxt("astro_image.csv")

plt.figure(figsize=(8,6))
plt.imshow(data, cmap="viridis")
plt.colorbar(label="Astronomical Image")
plt.show()

# 2)
plt.figure(figsize=(12,5))
zoom1 = data[1000:1200, 1000:1200]
zoom2 = data[1400:1600, 1400:1600]

plt.subplot(1, 2, 1)
plt.imshow(zoom1, cmap = "viridis")
plt.title("Zoom 1")

plt.subplot(1, 2, 2)
plt.imshow(zoom2, cmap = "viridis")
plt.title("Zoom 2")

plt.tight_layout()
plt.show()
```

```python
# -- 3.2 Pandas and Ramen -- 
ramen = pd.read_csv("ramen-ratings.csv")
print(ramen.head())

# 1)
# The column names are review #, brand, variety, style, country, stars, and top ten

# 2)
print(ramen.loc[26, "Style"])
# The ramen style is "Pack"

# 3)
print(ramen["Brand"].nunique())
# There are 355 different ramen brands

# 4)
ramen_sort = ramen.sort_values("Brand")
print(ramen_sort)

# 5)
ramen_USA = ramen[ramen["Country"] == "USA"]
print(ramen_USA["Brand"].nunique())
# There are 49 ramen brands made in the USA

# 6)
usa_pack = ramen[(ramen["Country"] == "USA") & (ramen["Style"] == "Pack")]
print(usa_pack["Brand"].nunique())
# 27 ramen brands in the USA make pack style ramen.

# 7)
top10 = ramen[ramen["Top Ten"].notna()]
top10_brands = top10["Brand"].unique()
print(top10_brands)
# ['MAMA' 'Prima Taste' 'Prima' 'Tseng Noodles' 'Wugudaochang', 'A-Sha Dry Noodle' 'MyKuali' 'CarJEN' 'Maruchan' 'Mamee' 'Mama', 'Sapporo Ichiban' 'Nongshim' 'Samyang Foods' 'Paldo' 'Indomie' 'Koka', 'Mi Sedaap' 'Nissin' 'Myojo' 'Doll']
```
